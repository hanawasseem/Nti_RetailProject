# **1.Revenue and Profit by Year**

```python
yearly = df1.groupby("sale_year").agg(
    revenue=("net_revenue", "sum"),
    profit=("total_profit", "sum")
)

yearly
```

# **2.Performance by Category**

```python
category = df1.groupby("category").agg(
    revenue=("net_revenue", "sum"),
    profit=("total_profit", "sum"),
    quantity=("quantity", "sum")
).sort_values("revenue", ascending=False)

category
```

# **3.Return Rate**

```python
return_rate = df1["returned"].mean() * 100

print(f"Overall Return Rate: {return_rate:.2f}%")
```

# **4.Return Rate By Category**

```python
category_return = (
    df1.groupby("category")["returned"]
    .mean()
    .mul(100)
    .sort_values(ascending=False)
)
```

# **5.Negative Profit**

```python
negative_profit = df1[df1["total_profit"] < 0]

print("Negative Profit Transactions:", len(negative_profit))
print("Total Loss:", negative_profit["total_profit"].sum())
```

# **6.Negative Profit by category**

```python
negative_by_category = (
    negative_profit
    .groupby("category")["total_profit"]
    .sum()
    .sort_values()
)

negative_by_category
```

# **7.Supplier Performance**

```python
supplier = df1.groupby("supplier").agg(
    revenue=("net_revenue", "sum"),
    profit=("total_profit", "sum"),
    quantity=("quantity", "sum")
).sort_values("profit", ascending=False)

supplier
```

# **8.Season Performance**

```python
season = df1.groupby("season").agg(
    revenue=("net_revenue", "sum"),
    profit=("total_profit", "sum")
).sort_values("revenue", ascending=False)

season
```

# **Merge customer with Product&Sales** **(age_group)**

```python
sales_customer = df1.merge(
    df2[["customer_id", "age_group"]],
    on="customer_id",
    how="left"
)
```

# **9.Age Group**

```python
age_analysis = sales_customer.groupby("age_group").agg(
    revenue=("net_revenue", "sum"),
    profit=("total_profit", "sum"),
    customers=("customer_id", "nunique")
).sort_values("revenue", ascending=False)

age_analysis
```

# **Merge customer with Product&Sales (city)**  

```python
sales_customer = df1.merge(
    df2[["customer_id", "city"]],
    on="customer_id",
    how="left"
)
```

# **10.Customer City**

```python
city_analysis = sales_customer.groupby("city").agg(
    revenue=("net_revenue", "sum"),
    profit=("total_profit", "sum")
).sort_values("revenue", ascending=False)

city_analysis
```

# **Visualization**

**Revenue and Profit by** **Year**

```python
yearly.plot(kind="bar", figsize=(10,5))

plt.title("Revenue and Profit by Year")
plt.xlabel("Year")
plt.ylabel("Amount")
plt.xticks(rotation=0)
plt.show()
```

**Revenue by Category**

```python
category["revenue"].plot(
    kind="bar",
    figsize=(10,5)
)

plt.title("Revenue by Category")
plt.xlabel("Category")
plt.ylabel("Revenue")
plt.xticks(rotation=45)
plt.show()
```

**Return Rate by Category**

```python
category_return.plot(
    kind="bar",
    figsize=(10,5)
)

plt.title("Return Rate by Category")
plt.xlabel("Category")
plt.ylabel("Return Rate (%)")
plt.xticks(rotation=45)
plt.show()
```

**Correlation Heatmap**

```python
import seaborn as sns
import matplotlib.pyplot as plt

# Correlation matrix
corr = df1.select_dtypes(include='number').corr()

# Plot
plt.figure(figsize=(14, 10))
sns.heatmap(
    corr,
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    center=0
)

plt.title("Correlation Heatmap")
plt.show()
```

 **Negative Profit Analysis**

* **8,378** transactions generated negative profit, resulting in approximately **496.5K** in total losses.

* This indicates that a significant number of transactions were not financially profitable, even when sales revenue was generated.

* The business should investigate the main reasons behind these negative-profit transactions, such as high product costs, low selling prices, or returned products.

* Identifying the products, categories, suppliers, and stores responsible for these losses can help the company reduce financial losses and improve profit margins.





# **Business Recommendations**

**1.** Focus on high-performing categories
    Prioritize Dresses and Accessories by increasing inventory and targeted marketing to maximize sales and profit.

**2.** Prepare for Summer demand
    Since Summer has the highest revenue and profit, the company should increase stock and marketing activities before and during this season.

**3.** Reduce product returns
    The overall return rate is **9.92%**. The company should identify the main causes of returns and improve sizing information, product descriptions, and quality control.

**4.** Reduce negative-profit transactions
    Investigate loss-making transactions and review pricing, product costs, and profit margins to minimize financial losses.
    
**5.** Optimize supplier and store performance
    Evaluate suppliers based on profitability, not revenue alone, and investigate stores with higher return rates to improve overall performance.

**6.** Target the most valuable customers
    Focus marketing efforts on the**25–45**age group, which represents a strong customer segment, and use customer data for more targeted promotions.
